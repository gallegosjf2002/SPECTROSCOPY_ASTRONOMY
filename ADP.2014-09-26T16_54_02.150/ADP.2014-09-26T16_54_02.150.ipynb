```python
import numpy as np

# 1. Cargar el espectro limpio (ajusta skiprows=1 si la primera fila tiene encabezados de texto)
datos = np.loadtxt("ADP_clean.txt")
wave = datos[:, 0]

# 2. Definir el corte en la brecha central (~5330 A)
corte = 5330.0

parte_azul = datos[wave <= corte]
parte_roja = datos[wave > corte]

# 3. Guardar ambos archivos en la misma carpeta
np.savetxt("ADP_clean_azul.txt", parte_azul, fmt="%.6f")
np.savetxt("ADP_clean_rojo.txt", parte_roja, fmt="%.6f")

print(f"Listo: Azul ({len(parte_azul)} pts) | Rojo ({len(parte_roja)} pts)")
```

```python
normPlot.normplot('ADP_clean_azul.txt')
```

```python
normPlot.normplot('ADP_clean_rojo.txt')
```
